```python
import pandas as pd

df = pd.read_csv("data/final_avalanche_data.csv")
df.shape
```

```python
# % missing (NaN) per column
missing_pct = (df.isna().mean() * 100).round(2).sort_values(ascending=False)
missing_pct[missing_pct > 0]
```

```python
# % of rows that have at least one missing value, and overall % of missing cells
rows_with_na_pct = (df.isna().any(axis=1).mean() * 100).round(2)
overall_pct = (df.isna().mean().mean() * 100).round(2)
print(f"rows with >=1 missing value: {rows_with_na_pct}%")
print(f"overall missing cells:       {overall_pct}%")
```
